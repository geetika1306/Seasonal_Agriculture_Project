```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

# Create cleaned dataset
clean_df = df.copy()

# Remove duplicate rows
clean_df = clean_df.drop_duplicates()

# Check the data
print("Dataset shape:", clean_df.shape)
print("Data loaded successfully!")
```

```python
season_yield = clean_df.groupby("Season")["Yield_Tonnes_Ha"].mean()

print(season_yield)

plt.figure(figsize=(8,5))
season_yield.plot(kind="bar")

plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=0)
plt.show()
```

```python
season_yield = clean_df.groupby("Season")["Yield_Tonnes_Ha"].mean()

plt.figure(figsize=(8,5))

sns.barplot(
    x=season_yield.index,
    y=season_yield.values
)

plt.title("Average Agricultural Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")

plt.tight_layout()
plt.show()
```

```python
season_profit = clean_df.groupby("Season")["Profit_INR"].mean()

plt.figure(figsize=(8,5))

sns.barplot(
    x=season_profit.index,
    y=season_profit.values
)

plt.title("Average Agricultural Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=clean_df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Rainfall vs Agricultural Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=clean_df,
    x="Avg_Temperature_C",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Temperature vs Agricultural Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Yield (Tonnes/Ha)")

plt.tight_layout()
plt.show()
```

```python
irrigation_yield = clean_df.groupby(
    "Irrigation_Method"
)["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)

plt.figure(figsize=(9,5))

sns.barplot(
    x=irrigation_yield.index,
    y=irrigation_yield.values
)

plt.title("Average Yield by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Average Yield (Tonnes/Ha)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
crop_season_table = clean_df.pivot_table(
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

plt.figure(figsize=(10,7))

sns.heatmap(
    crop_season_table,
    annot=True,
    fmt=".2f"
)

plt.title("Average Crop Yield Across Seasons")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()
```

```python
# Average yield by season
season_yield = clean_df.groupby("Season")["Yield_Tonnes_Ha"].mean()

best_yield_season = season_yield.idxmax()
best_yield_value = season_yield.max()

lowest_yield_season = season_yield.idxmin()
lowest_yield_value = season_yield.min()

print("Best season for yield:", best_yield_season)
print("Average yield:", round(best_yield_value, 2))

print("\nLowest season for yield:", lowest_yield_season)
print("Average yield:", round(lowest_yield_value, 2))
```

```python
season_profit = clean_df.groupby("Season")["Profit_INR"].mean()

best_profit_season = season_profit.idxmax()
best_profit_value = season_profit.max()

lowest_profit_season = season_profit.idxmin()
lowest_profit_value = season_profit.min()

print("Best season for profit:", best_profit_season)
print("Average profit:", round(best_profit_value, 2))

print("\nLowest season for profit:", lowest_profit_season)
print("Average profit:", round(lowest_profit_value, 2))
```

```python
season_water_efficiency = clean_df.groupby("Season")["Water_Efficiency_t_per_1000m3"].mean()

best_water_season = season_water_efficiency.idxmax()
best_water_value = season_water_efficiency.max()

print("Most water-efficient season:", best_water_season)
print("Water efficiency:", round(best_water_value, 2))
```

```python
crop_season = clean_df.groupby(
    ["Season", "Crop"]
)["Yield_Tonnes_Ha"].mean().reset_index()

best_crop_each_season = crop_season.loc[
    crop_season.groupby("Season")["Yield_Tonnes_Ha"].idxmax()
]

print("Best crop in each season:")
print(best_crop_each_season)
```

```python
environmental_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct"
]

environmental_corr = clean_df[
    environmental_columns + ["Yield_Tonnes_Ha"]
].corr()["Yield_Tonnes_Ha"].drop("Yield_Tonnes_Ha")

print("Relationship between environmental factors and yield:")
print(environmental_corr.sort_values(ascending=False))
```

```python
risk_yield_corr = clean_df[
    ["Disease_Pest_Risk_pct", "Yield_Tonnes_Ha"]
].corr().iloc[0, 1]

print("Correlation between disease/pest risk and yield:",
      round(risk_yield_corr, 3))
```

```python
final_summary = clean_df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean",
    "Revenue_INR": "mean",
    "Total_Cost_INR": "mean",
    "Rainfall_mm": "mean",
    "Avg_Temperature_C": "mean",
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Disease_Pest_Risk_pct": "mean"
}).round(2)

print(final_summary)
```
